# Exploratory Data Analysis — Weather Data Warehouse

This notebook queries the SQLite database created by the ETL pipeline to analyze temperature trends, city comparisons, and missing data.

```python
import sqlite3
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Connect to warehouse database
conn = sqlite3.connect('../data/warehouse.db')

# Query daily summary view
df_daily = pd.read_sql_query('SELECT * FROM vw_daily_city_summary', conn)
df_daily.head()
```

```python
# Plot average temperature comparison across cities
plt.figure(figsize=(10, 5))
sns.barplot(data=df_daily, x='city_name', y='avg_temp_c', palette='viridis')
plt.title('Average Temperature by City (°C)')
plt.xlabel('City')
plt.ylabel('Avg Temp (°C)')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
# Query hourly temperature time-series
df_hourly = pd.read_sql_query("""
    SELECT f.timestamp_utc, c.city_name, f.temperature_c, f.relative_humidity
    FROM fact_weather_metrics f
    JOIN dim_city c ON f.city_id = c.city_id
    ORDER BY f.timestamp_utc ASC
""", conn)

df_hourly['timestamp_utc'] = pd.to_datetime(df_hourly['timestamp_utc'])

plt.figure(figsize=(12, 6))
sns.lineplot(data=df_hourly, x='timestamp_utc', y='temperature_c', hue='city_name', marker='o')
plt.title('Hourly Temperature Trend by City')
plt.xlabel('Timestamp (UTC)')
plt.ylabel('Temperature (°C)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Close database connection
conn.close()
```
